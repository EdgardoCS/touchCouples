```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

import statsmodels.formula.api as smf
```

```python
df = pd.read_csv('source/csv/FReDA4.csv')
df = df[df["Role"] == "Anchor"].dropna()
```

```python
o = df['Relationship Length (years)'].quantile(0.15)
p = df['Relationship Length (years)'].quantile(0.50)
q = df['Relationship Length (years)'].quantile(0.75)
```

```python
def len_to_bins(val):
    if val <= 2.5:
        return "Short (2.5 years)"
    elif val <= 10:
        return "Medium (2.6 to 10 years)"
    else:
        return "Long (10+ years)"

df['Relationship Length (bins)'] = df['Relationship Length (years)'].apply(len_to_bins)
```

```python
pd.crosstab(df["Group3"], df["Relationship Length"])
```

```python
df.rename(columns={'Relationship Length (years)':'Relationship_length_years'}, inplace=True)
```

### *Frequency*

```python
model = smf.ols('Frequency ~ Age * Relationship_length_years', data=df).fit()

print(model.summary())
```

```python
pal1 = [
    '#C87A5E',
    '#6D947B',
    '#84739E'
]

pal2 = [
    '#5D8AA8',
    '#D4A373',
    '#C0778B'
]

pal3 = [
    '#E88D6F',
    '#73B5B3',
    '#7B85B5'
]

pal4 = [
    '#87BBA2',
    '#F3C677',
    '#5C80A1'
]

order = [
    "Short (2.5 years)",
    "Medium (2.6 to 10 years)",
    "Long (10+ years)"
]

font_axis = 32
font_ticks = 24
font_legend = 16

label_distance = 5

bins = 1500
```

```python
pal = pal2
```

```python
g = sns.lmplot(
    data=df,
    x="Age",
    x_ci=None,
    x_bins=bins,
    y="Frequency",
    hue="Relationship Length (bins)",
    hue_order=order,
    palette=pal,
    height=6,
    aspect=1.5,

    # logx=True,
)

plt.xlabel("Age", fontsize=font_axis-4)
plt.ylabel("Frequency of Touch", fontsize=font_axis, labelpad = label_distance)
plt.xticks(fontsize=font_ticks)

plt.yticks([0, 1, 2, 3, 4, 5],
           labels=['Not at all', '1-5\ntimes', '6-10\ntimes', '11-20\ntimes', '21-50\ntimes',
                   'More than\n 50 times'],
           rotation=0,
           fontsize=font_ticks
           );

plt.grid(False)

g._legend.set_title("Relationship Length")

sns.move_legend(g, loc="center", bbox_to_anchor=(.91, 0.5))
plt.setp(g._legend.get_texts(), fontsize=font_legend)
plt.setp(g._legend.get_title(), fontsize=font_legend)

plt.ylim([0, 5.5])

plt.savefig('output/img/time/age_frequency.png', dpi=600, transparent=False, bbox_inches='tight')

plt.show()
```
