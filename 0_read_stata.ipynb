```python
import pandas as pd
```

```python
data = pd.read_stata("source/stata/W2B_com.dta")
```

```python
reldur = data[["reldur_w2b", "age_w2b", "page_w2b"]]
```

```python
reldur = reldur[reldur["reldur_w2b"]!= "Does not apply (filter)"]
reldur = reldur[reldur["reldur_w2b"]!= "Incomplete data"]
```

```python
reldur["reldur_w2b"] = pd.to_numeric(reldur["reldur_w2b"], errors="coerce")
reldur["years"] = (reldur["reldur_w2b"]) / 12
```
